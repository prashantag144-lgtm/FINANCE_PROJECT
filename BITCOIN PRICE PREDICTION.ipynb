```python
df=df.sort_values('Date',ascending=True)
```

```python
df.head()
```

```python
df = df.reset_index()
```

```python
df.head()
```

```python
df.reset_index(drop=True)
```

```python
df['Close_Lag1'] = df['Close'].shift(1)
```

```python
df['Close_Lag2']=df['Close'].shift(2)
df['Close_Lag3']=df['Close'].shift(3)
df['Close_Lag4']=df['Close'].shift(4)
df['Close_Lag7']=df['Close'].shift(7)
df['Close_Lag14']=df['Close'].shift(14)
```

```python
df['Close_MA7']=df['Close'].rolling(window=7).mean()
df['Close_MA30']=df['Close'].rolling(window=30).mean()
```

```python
df['Volatility_7']=df['Close'].rolling(window=7).std()
df['Volatility_30']=df['Close'].rolling(window=30).std()
```

```python
df['Returns']=df['Close'].pct_change()
```

```python
df.head()
```

```python
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python

X = df[['Close_Lag1', 'Close_Lag2','Close_Lag3','Close_Lag4','Close_Lag7','Close_Lag14', 'Close_MA7', 'Close_MA30',
        'Volatility_7', 'Volatility_30','Returns']]
y=df[['Close']]
```

```python
df.columns
```

```python
split=int(len(df)*0.8)
```

```python
X_train=X[:split]
y_train=y[:split]
```

```python
X_test=X[split:]
y_test=y[split:]
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_test.shape
```

```python
y_test.shape
```

```python
y
```

```python
X
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test,y_pred)
```
